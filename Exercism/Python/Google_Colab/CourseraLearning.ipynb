```python
hrs = float(input("Enter Hours:"))
rate = float(input("Enter Rate:"))
if hrs <= 40:
  pay = hrs * rate
else:
  pay = (40 * rate) + ((hrs - 40) * rate * 1.5)
print(pay)
```

```python
try:
  score = float(input("Enter Score: "))
  if score > 1.0 or score < 0.0:
    print("Error, score must be between 0.0 and 1.0")
  elif 1.0 > score >= 0.9:
    print("A")
  elif score >= 0.8:
    print("B")
  elif score >= 0.7:
    print("C")
  elif score >= 0.6:
    print("D")
  elif score < 0.6:
    print("F")
except ValueError as e:
  print(f'Error is based on type of: {type(e).__name__}')
```
