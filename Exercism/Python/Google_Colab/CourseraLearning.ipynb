```python
x = [1, 2, 3]
y = x
x = [4, 5, 6]
print("Cuando reasignamos 'x' a un nuevo objeto:")
print(f"  x apunta a: {x}")
print(f"  y sigue apuntando a: {y}\n")

print("--------------------------------------------------\n")

# En este caso, 'y' reflejará los cambios porque se modifica el objeto al que ambas variables referencian.
z = [10, 20, 30]
w = z
print("Inicialmente, z y w referencian el mismo objeto:")
print(f"  z: {z}")
print(f"  w: {w}\n")

# Modificamos el objeto al que 'z' (y 'w') referencia
z.append(40)
print("Después de modificar el objeto al que 'z' referencia (z.append(40)):")
print(f"  z ahora es: {z}")
print(f"  w también es: {w}\n")

# Para verificar si son el mismo objeto, podemos usar id()
print("Direcciones de memoria (IDs) de los objetos:")
print(f"  id de z: {id(z)}")
print(f"  id de w: {id(w)}")

z.append(90)
print("Después de modificar el objeto al que 'z' referencia (z.append(40)):")
print(f"  z ahora es: {z}")
print(f"  w también es: {w}\n")

z = [10, 20, 30, 40, 90]
print(f"  z ahora es: {z}")
```

```python
def mi_decorador(func):
  def mi_wrapper(*args, **kwargs):
    print("Antes de la ejecución de la función")
    print(f"Los argumentos de entrada son: {args} y {kwargs}")
    result = func(*args, **kwargs)
    print("Después de la ejecución de la función")
    return result
  return mi_wrapper

@mi_decorador
def suma(a: int, b: int) -> float:
  print(f"Dentro de la función suma: {a + b}")


suma(a="5", b="10.09")
```

```python
# ✅ BIEN: Nombres expresivos, autodocumentados y alineados a PEP 8.

class ProcesadorPedidos:
    """Gestiona la validación de estado y el cálculo de importes para pedidos."""

    TASA_IMPUESTO: float = 0.16
    PORCENTAJE_DESCUENTO: float = 0.10
    UMBRAL_MONTO_DESCUENTO: float = 1000.0

    def calcular_totales_pedidos(self, lista_pedidos: list[dict]) -> list[float]:
        """Calcula el total a pagar para una lista de pedidos con estado activo."""
        totales_procesados: list[float] = []

        for pedido in lista_pedidos:
            if pedido.get("estado") == "ACTIVO":
                monto_base = pedido["monto"]
                monto_con_impuesto = monto_base * (1 + self.TASA_IMPUESTO)

                if self.es_elegible_para_descuento(
                    monto_base,
                    es_cliente_vip=pedido.get("es_vip", False)
                ):
                    monto_con_impuesto *= (1 - self.PORCENTAJE_DESCUENTO)

                totales_procesados.append(round(monto_con_impuesto, 2))

        return totales_procesados

    def es_elegible_para_descuento(self, monto_base: float, es_cliente_vip: bool) -> bool:
        """Determina si un pedido califica para aplicar un descuento."""
        return monto_base > self.UMBRAL_MONTO_DESCUENTO or es_cliente_vip


# Uso claro y legible para cualquier desarrollador
procesador = ProcesadorPedidos()

pedidos_ejemplo = [
    {"estado": "ACTIVO", "monto": 1200.0, "es_vip": False},
    {"estado": "INACTIVO", "monto": 500.0, "es_vip": True},
    {"estado": "ACTIVO", "monto": 800.0, "es_vip": True},
]

totales = procesador.calcular_totales_pedidos(pedidos_ejemplo)
print(f"Totales finales calculados: {totales}")
```
