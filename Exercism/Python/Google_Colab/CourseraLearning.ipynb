```python
def mi_decorador(func):
  def mi_wrapper(*args, **kwargs):
    print("Antes de la ejecución de la función")
    print(f"Los argumentos de entrada son: {args} y {kwargs}")
    result = func(*args, **kwargs)
    print("Después de la ejecución de la función")
    return result
  return mi_wrapper

@mi_decorador
def suma(a: int, b: int) -> float:
  print(f"Dentro de la función suma: {a + b}")


suma(a="5", b="10.09")
```

```python
# ✅ BIEN: Nombres expresivos, autodocumentados y alineados a PEP 8.

class ProcesadorPedidos:
    """Gestiona la validación de estado y el cálculo de importes para pedidos."""

    TASA_IMPUESTO: float = 0.16
    PORCENTAJE_DESCUENTO: float = 0.10
    UMBRAL_MONTO_DESCUENTO: float = 1000.0

    def calcular_totales_pedidos(self, lista_pedidos: list[dict]) -> list[float]:
        """Calcula el total a pagar para una lista de pedidos con estado activo."""
        totales_procesados: list[float] = []

        for pedido in lista_pedidos:
            if pedido.get("estado") == "ACTIVO":
                monto_base = pedido["monto"]
                monto_con_impuesto = monto_base * (1 + self.TASA_IMPUESTO)

                if self.es_elegible_para_descuento(
                    monto_base,
                    es_cliente_vip=pedido.get("es_vip", False)
                ):
                    monto_con_impuesto *= (1 - self.PORCENTAJE_DESCUENTO)

                totales_procesados.append(round(monto_con_impuesto, 2))

        return totales_procesados

    def es_elegible_para_descuento(self, monto_base: float, es_cliente_vip: bool) -> bool:
        """Determina si un pedido califica para aplicar un descuento."""
        return monto_base > self.UMBRAL_MONTO_DESCUENTO or es_cliente_vip


# Uso claro y legible para cualquier desarrollador
procesador = ProcesadorPedidos()

pedidos_ejemplo = [
    {"estado": "ACTIVO", "monto": 1200.0, "es_vip": False},
    {"estado": "INACTIVO", "monto": 500.0, "es_vip": True},
    {"estado": "ACTIVO", "monto": 800.0, "es_vip": True},
]

totales = procesador.calcular_totales_pedidos(pedidos_ejemplo)
print(f"Totales finales calculados: {totales}")
```

Ejemplo con clases y el punto importante aquí es el uso del decorador cached_property que está comentado porque la función summary no acepta parámetros con ese decorador.

Info de Gemini:
Entonces, la clave es recordar que @cached_property convierte el método en un atributo de solo lectura (después de la primera evaluación) al que se accede sin paréntesis, al igual que accederías a report.calls o report.result.


```python
from functools import cached_property

class Report:
  def __init__(self) -> None:
     self.calls = 0
     self.result = 0


  def summary(self, a: int, b: int) -> float:
    self.calls += 1
    self.result += a + b
    return self.result

  @cached_property
  def summary2(self):
    self.calls += 1
    return "Done"


"""@cached_property"""
# Outside class
report = Report()
report.summary(5, 5)
report.summary(30, 50)

print(report.calls)
print(report.result)

# Importante, con el wrapper se quitan los parentesis
report.summary2
report.summary2
print(report.calls)

```

Clase con un constructor que permite mútiples argumentos de entrada y hace un check de dichas entradas




> *args, **kwargs

args es lista

kwards es diccionario

```python
class Cosa:
    def __init__(self, *args, **kwargs):
      self.nombre = None
      self.edad = None
      if 'nombre' in kwargs:
          self.nombre = kwargs['nombre']
      elif args:
          self.nombre = args[0]
      else:
          self.nombre = "Desconocido"

      if 'edad' in kwargs:
          self.edad = kwargs['edad']
      elif args:
          self.edad = args[1]
      else:
          self.edad = 0


nueva_cosa = Cosa(nombre="Daniel", edad=30)
print(nueva_cosa.nombre)
print(f"Y su edad es: {nueva_cosa.edad}")
```

# Reading Files (Uner "content" directory)

`open()` - connection to the file (not the storage)

`read()` - todo en un solo string pegado en una línea

```python
from os.path import join
import sys

file_name = "my_text_file.txt"  # corregir este nombre para que continue con
file_path = join("/content/sample_data/", file_name)
try:
  file_handler = open(file_path, 'r')
except Exception as e:
  print(f"No pude abrir el archivo seleccionado: {e}")
else:
  count = 0
  for line in file_handler:
    line = line.rstrip()
    count += 1
    line = line.upper()
    print(line)


  print(f"Total lines: {count}")
  file_handler.close()
```
